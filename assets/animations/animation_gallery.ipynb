## Backpropagation through several updates

States move forward through two update maps. Adjoint sensitivities move backward and each use of the shared parameter contributes to its gradient.

**Scope:** Computational-graph schematic. Assumes differentiable maps, fixed initial state and a final-state-only loss.

**Question:** Why does the parameter gradient contain a contribution from each update?

```python
display(Video(str(animation_root / 'learning/reverse_accumulation.mp4'), embed=True, width=960, html_attributes='controls playsinline'))
```

<details><summary>Worked answer</summary>

The same parameter enters both maps. The chain rule adds both paths to the loss. Here the initial state is fixed and the loss depends on the final state. Parameter-dependent initial states or explicit parameter dependence in the loss contribute additional terms.

</details>

**Animation description**

1. Specify a fixed initial state and a shared parameter.
2. Evaluate two differentiable update maps and a scalar final loss.
3. Propagate the final-state adjoint backward.
4. Sum the parameter contribution from every update.
5. An optimiser chooses a separate update using the resulting gradient.
